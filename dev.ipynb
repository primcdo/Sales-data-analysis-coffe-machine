Pode ser a diferença de tamanho talvez?

```python
# Quais preços aparecem associados a cada bebida?
tabela_precos = (
    data_1.groupby("coffee_name")["money"]
    .value_counts()
    .unstack(fill_value=0)
)

display(tabela_precos)
```

```python
# Converter as colunas para o formato correto
data_1["date"] = pd.to_datetime(data_1["date"])
data_1["datetime"] = pd.to_datetime(data_1["datetime"])

# Ordenar as vendas por data
df = data_1.sort_values("datetime")

# Exibir as primeiras e as últimas transações
display(df[["datetime", "coffee_name", "money"]].head(10))
display(df[["datetime", "coffee_name", "money"]].tail(10))
```

É, provavelmente o preço mudou, mas como não há informação sobre tamanho e os valores de um ano de diferença são bem similares, é possível que tenha sido apenas um reajuste.

```python
# Criar uma coluna com o mês e o ano
df["mes"] = df["datetime"].dt.to_period("M")

# Ver quantos preços diferentes cada bebida teve em cada mês
precos_mensais = (
    df.groupby(["mes", "coffee_name"])["money"]
    .agg(["min", "max", "nunique"])
    .reset_index()
)

display(precos_mensais)
```
